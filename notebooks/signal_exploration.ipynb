# WESAD Signal Exploration

## Research Objective

This notebook explores the physiological signals in the WESAD dataset to understand their characteristics, quality, temporal behavior, and variation across subjects.

The goal is to inform subsequent preprocessing and experimental design decisions for a physiological stress classification task.

## Scope

Exploratory analysis of the Empatica E4 wrist-worn sensor modalities:

- BVP — Blood Volume Pulse (64 Hz)
- ACC — Accelerometer (32 Hz)
- EDA — Electrodermal Activity (4 Hz)
- TEMP — Skin Temperature (4 Hz)
